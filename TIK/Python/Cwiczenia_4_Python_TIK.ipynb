## NumPy: Numeryczny Python

NumPy zapewnia wydajny sposób przechowywania i przetwarzania wielowymiarowych tablic gęstych w języku Python.
Najważniejsze cechy NumPy to:

- Udostępnia strukturę ``ndarray``, która umożliwia wydajne przechowywanie i przetwarzanie wektorów, macierzy oraz zbiorów danych o wyższych wymiarach.
- Zapewnia czytelny i wydajny składniowy sposób operowania na tych danych, od prostych operacji arytmetycznych na poszczególnych elementach po bardziej skomplikowane operacje algebry liniowej.

W najprostszym przypadku tablice NumPy wyglądają bardzo podobnie do list w Pythonie.
Na przykład, oto tablica zawierająca zakres liczb od 1 do 9 (porównaj to z wbudowaną funkcją Pythona ``range()``):

````python
import numpy as np
x = np.arange(1, 10)
x
````

````python
y = list(range(1,10))
y
````

````python
print(x, y)
````

Tablice w NumPy zapewniają zarówno wydajne przechowywanie danych, jak i wydajne operacje na poszczególnych elementach danych.
Na przykład, aby podnieść każdy element tablicy do kwadratu, możemy zastosować operator ``**`` bezpośrednio do tablicy:

````python
%%time
x ** 2
````

Porównajmy to z dużo bardziej rozbudowanym wyrażeniem listowym w stylu Pythona, dającym ten sam wynik:

````python
%%time
[val ** 2 for val in range(1, 10)]
````

W przeciwieństwie do list w języku Python (które są jednowymiarowe), tablice NumPy mogą być wielowymiarowe.
Na przykład, w tym miejscu zmienimy kształt naszej tablicy ``x`` na tablicę o wymiarach 3x3:

````python
M = x.reshape((3, 3))
M
````

Tablica dwuwymiarowa jest jednym ze sposobów przedstawienia macierzy, a biblioteka NumPy potrafi sprawnie wykonywać typowe operacje macierzowe. Na przykład transpozycję można obliczyć za pomocą operatora ``.T``:

````python
M.T
````

lub iloczyn macierzowo-wektorowy przy użyciu funkcji ``np.dot``:

````python
np.dot(M, [5, 6, 7])
````

a nawet bardziej zaawansowane operacje, takie jak rozkład na wartości własne:

````python
np.linalg.eigvals(M)
````

Takie operacje z zakresu algebry liniowej stanowią podstawę znacznej części współczesnej analizy danych, zwłaszcza w dziedzinach uczenia maszynowego i eksploracji danych.

## SciPy: Naukowy Python

SciPy to zbiór funkcji naukowych oparty na bibliotece NumPy.
Pakiet ten powstał jako zestaw nakładek w języku Python na znane biblioteki Fortran służące do obliczeń numerycznych, a następnie rozrósł się.
Pakiet ma strukturę zestawu podmodułów, z których każdy implementuje pewną klasę algorytmów numerycznych.
Oto niekompletna lista niektórych z ważniejszych modułów dla nauki o danych:

- ``scipy.fftpack``: Szybkie transformacje Fouriera
- ``scipy.integrate``: Integracja numeryczna
- ``scipy.interpolate``: Interpolacja numeryczna
- ``scipy.linalg``: procedury algebry liniowej
- ``scipy.optimize``: numeryczna optymalizacja funkcji
- ``scipy.sparse``: przechowywanie macierzy rzadkich i algebra liniowa
- ``scipy.stats``: procedury analizy statystycznej

Na przykład, przyjrzyjmy się interpolacji gładkiej krzywej pomiędzy pewnymi danymi

````python
from scipy import interpolate

# choose eight points between 0 and 10
x = np.linspace(0, 10, 8)
y = np.sin(x)

# create a cubic interpolation function
func = interpolate.interp1d(x, y, kind='cubic')

# interpolate on a grid of 1,000 points
x_interp = np.linspace(0, 10, 1000)
y_interp = func(x_interp)

# plot the results
plt.figure()  # new figure
plt.plot(x, y, 'o')
plt.plot(x_interp, y_interp);
````

Widzimy płynną interpolację między punktami.

## Przykład dopasowania prostej

W poprzedniej części poznaliśmy przydatne narzędzia do analizy danych oraz wykorzystaliśmy interpolacje z biblioteki ``scipy.optimize``. Przy analizie danych zazwyczaj znamy prawa stojące za wewnętrzną strukturą danych, ale również zdarza się, że musimy zaproponować nową hipotezę do opisu ich kształtu. W obu tych przypadkach będziemy potrzebować znaleźć parametry funkcji (***popt***) oraz niepewności ich wyznaczenia (obliczane z ***pcov***).
<br>
<br>
Zacznijmy od zaimportowania ``scipy.optimize.curve_fit`` i ustalenia ziarna (z j. ang. *seed*) w celu uzyskania spójnych wyników.

````python
from scipy.optimize import curve_fit
np.random.seed(42)
````

W kolejnym etapie spreparujemy syntetyczne dane, których użyjemy do dopasowania funkcji. Nie trzeba dodawać (ale nie mogłem się powstrzymać i dodałem), że zazwyczaj dane zwracane są przez aparaturę i nie znamy szczegółów ich tworzenia.
<br>
<br>
Współrzędna ``x`` to dziedzina funkcji, dlatego standardową praktyką jest przyjęcie rozkładu liniowego o stałym kroku, np. przy pomocy funkcji ``numpy.linspace``.
<br>
<br>
Współrzędna ``y`` to wartości, które symulujemy. Posłużymy się w tym przykładzie zwykłą zależnością liniową, do której dodajemy szum w postaci losowych wartości z rozkładu płaskiego - ``numpy.random.uniform()``.  

````python
x = np.linspace(0, 10, 16)
y = -0.102 * x + np.random.uniform(-0.1, 0.1, size=len(x))
````

W programowaniu złotą zasadą jest by oglądać wynik wykonywania swojego kodu przy możliwie małej iteracji zmian. Każdorazowo dodając nową funkcjonalność właściwym odruchem jest postawienie hipotezy - *jak wpłynie ta funkcjonalność na wynik?* W komórce wyżej wygenerowaliśmy dane, a w poprzedniej części nauczyliśmy się jak możemy je narysować na wykresie przy użyciu ``matplotlib.pyplot``.

````python
plt.figure()
plt.plot(x, y, 'o')
plt.show()
````

Ewidentnie widzimy liniową tendencję spadkową, choć losowy szum jest na tyle duży, że można go pomylić ze strukturą danych. Niemniej jednak widzimy, że stworzylśmy dane o stałym kroku, w oczekiwanym przedziale, z zależnością liniową i szumem.
<br>
<br>
Powyższym danym można nadać (zmyśloną) interpretację fizyczną. Wyobraźmy sobie, że mamy do czynienia z opróżnianiem zbiornika na wodę o liniowym tempie ``a``, które ma początkową objętość wody ``b``. Każdy fizyczny pomiar obarczony jest błędem, stąd odchylenia od postawionej hipotezy. Niemniej jednak spróbujmy uzyskać tempo odpływu wody ``a``. W tym celu zdefiniujmy funkcję:

````python
def func(x, a, b):
    return a * x + b
````

Będę konsekwentny we własnej zasadzie, którą dla niepoznaki nazwałem złotą, i narysuję tę zależność w sposób naiwny. Widzę, że wykres zaczyna się w okolicach ``(0,0)`` i przechodzi przez ``(8,0.8)``. Zgadnę zatem wartości ``a, b`` i zobaczę czy faktycznie ta hipoteza ma sens. Skorzystam do tego z pakietu ``matplotlib.pyplot.subplots``, który był omawiany, jak wszyscy pamiętają, na wykładzie:

````python
zgadnięte_a = -0.1
zgadnięte_b = 0.0

from matplotlib.pyplot import subplots
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

axes[0].scatter(x, y)
axes[1].plot(x, func(x, zgadnięte_a, zgadnięte_b))
axes[2].scatter(x, y)
axes[2].plot(x, func(x, zgadnięte_a, zgadnięte_b))
````

Hipoteza wygląda sensownie, zatem przejdźmy do dopasowania funkcji. Wykorzystam w tym celu poprzednie zgadnięte wartości jako wartości początkowe, od których algorytm zacznie regularyzacje (na potrzebę tego kursu nie trzeba się zagłębiać jak to działa oraz dlaczego to działa). Wykorzystam w tym celu funkcję ``curve_fit`` ([dokumentacja](https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.curve_fit.html)).

````python
initial_guess = [zgadnięte_a, zgadnięte_b]
popt, pcov = curve_fit(func, x, y, p0=initial_guess) # Funkcja zwraca parametry oraz ich kowariancje
perr = np.sqrt(np.diag(pcov))
````

Na tym etapie możemy wstawić nasze parametry do funkcji, tak jak robiliśmy to wcześniej przy samym zgadnięciu:

````python
fig, axes = plt.subplots(1, 1, figsize=(6, 5))
#Zauważ jak zmienia się figsize gdy rozmiar przechodzi z (1,3) na (1,1).
#Dobrą praktyką jest trzymanie się konwencji: plt.subplots(a, b, figsize=(6*a, 5*b))

axes.scatter(x, y, label='Dane')
axes.plot(x, func(x, *popt), 'r-', label='Dopasowanie')
axes.legend()
axes.set_title("Wynik dopasowania")
````

Zobaczmy teraz jaką wartość szukanych parametrów ``a,b`` uzyskaliśmy z dopasowania:

````python
print("Wyniki dopasowania")
param_names = ['a', 'b']
for name, value, error in zip(param_names, popt, perr):
    print(f"Parametr {name}: {value:10.4f} ± {error:.4f}")
````
