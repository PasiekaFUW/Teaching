# Iteratory

Często istotnym elementem analizy danych jest wielokrotne, zautomatyzowane powtarzanie podobnych obliczeń.
Możesz na przykład mieć tabelę zawierającą nazwiska, które chcesz podzielić na imiona i nazwiska, albo daty, które chcesz przekonwertować do jakiegoś standardowego formatu.
Jednym z rozwiązań oferowanych przez Pythona jest składnia *iteratora*.
Widzieliśmy to już na przykładzie iteratora ``range``:

````python
for i in range(10):
    print(i, end=' ')
````

W tym miejscu zajrzymy nieco głębiej. Okazuje się, że w Pythonie 3 obiekt `range` nie jest listą, lecz tzw. iteratorem, a poznanie jego działania jest kluczem do zrozumienia szerokiej gamy bardzo przydatnych funkcji języka Python.

## Iterowanie przez listy
Najłatwiej zrozumieć iteratory na konkretnym przykładzie przeglądania listy. Rozważmy następujący przykład:

````python
for value in [2, 4, 6, 8, 10]:
    # do some operation
    print(value + 1, end=' ')
````

Znana składnia „``for x in y``” pozwala nam powtarzać pewną operację dla każdej wartości z listy.
Fakt, że składnia kodu jest tak zbliżona do jego angielskiego opisu („*for [each] value in [the] list*”), to tylko jeden z wielu zabiegów składniowych, które sprawiają, że Python jest językiem tak intuicyjnym w nauce i użyciu.

Jednak to, co widać na pierwszy rzut oka, nie jest tym, co *naprawdę* się dzieje.
Kiedy piszesz coś w rodzaju „``for val in L``”, interpreter Pythona sprawdza, czy posiada on interfejs *iteratora*, co możesz sprawdzić samodzielnie za pomocą wbudowanej funkcji ``iter``:

````python
iter([2, 4, 6, 8, 10])
````

To właśnie ten obiekt iteratora zapewnia funkcjonalność wymaganą przez pętlę „for”.
Obiekt „iter” jest pojemnikiem, który zapewnia dostęp do kolejnego obiektu tak długo, jak długo jest on ważny, co można zaobserwować za pomocą wbudowanej funkcji „next”:

````python
I = iter([2, 4, 6, 8, 10])
````

````python
print(next(I))
````

````python
print(next(I))
````

````python
print(next(I))
````

Jaki jest cel takiego stopnia pośrednictwa?
Okazuje się, że jest to niezwykle przydatne, ponieważ pozwala Pythonowi traktować jako listy elementy, które *w rzeczywistości nie są listami*.

## ``range()``: Lista nie zawsze jest listą
Być może najczęstszym przykładem tego rodzaju pośredniego iterowania jest funkcja ``range()`` w Pythonie 3 (w Pythonie 2 nazywana ``xrange()``), która zwraca nie listę, lecz specjalny obiekt ``range()``:

````python
range(10)
````

``range``, podobnie jak lista, udostępnia iterator:

````python
iter(range(10))
````

Więc Python wie by traktować ją *tak jakby* była listą:

````python
for i in range(10):
    print(i, end=' ')
````

Zaletą pośrednictwa iteratora jest to, że *cała lista nigdy nie jest tworzona wprost!*
Możemy to zaobserwować, wykonując obliczenia z wykorzystaniem funkcji `range`, które przeciążyłyby pamięć systemu, gdybyśmy faktycznie utworzyli listę (należy pamiętać, że w Pythonie 2 funkcja `range` tworzy listę, więc uruchomienie poniższego kodu nie przyniesie dobrych skutków!):

````python
N = 10 ** 12
for i in range(N):
    if i >= 10: break
    print(i, end=', ')
````

Gdyby funkcja ``range`` faktycznie utworzyła listę zawierającą bilion wartości, zajęłaby ona dziesiątki terabajtów pamięci komputera: byłaby to strata, biorąc pod uwagę, że ignorujemy wszystkie wartości poza pierwszymi dziesięcioma!

W rzeczywistości nie ma żadnego powodu, dla którego iteratory miałyby w ogóle się kończyć!
Biblioteka ``itertools`` języka Python zawiera funkcję ``count``, która działa jak nieskończony zakres:

````python
from itertools import count

for i in count():
    if i >= 10:
        break
    print(i, end=', ')
````

Gdybyśmy nie dodali tutaj warunku przerwania pętli, program kontynuowałby liczenie, dopóki proces nie zostałby ręcznie przerwany lub zakończony (na przykład za pomocą skrótu klawiszowego ``ctrl-C``).

## Przydatne iteratory
Ta składnia iteratorów jest stosowana niemal powszechnie zarówno w typach wbudowanych języka Python, jak i w obiektach bardziej związanych z nauką o danych, które omówimy w kolejnych rozdziałach.
W tym miejscu omówimy niektóre z bardziej przydatnych iteratorów w języku Python:

### ``enumerate``
Często zdarza się, że trzeba nie tylko przeglądać wartości w tablicy, ale także śledzić indeks.
Można ulec pokusie, by zrobić to w następujący sposób:

````python
L = [2, 4, 6, 8, 10]
for i in range(len(L)):
    print(i, L[i])
````

Chociaż to rozwiązanie działa, Python oferuje bardziej przejrzystą składnię dzięki iteratorowi `enumerate`:

````python
for i, val in enumerate(L):
    print(i, val)
````

To jest bardziej „pythoniczny” sposób na wyliczenie indeksów i wartości na liście.

### ``zip``
Czasami zdarza się, że chcesz jednocześnie przeglądać wiele list.
Oczywiście można przeglądać indeks tak jak w nie-pythonicznym przykładzie, który omawialiśmy wcześniej, ale lepiej jest użyć iteratora ``zip``, który łączy obiekty iterowalne:

````python
L = [2, 4, 6, 8, 10]
R = [3, 6, 9, 12, 15]
for lval, rval in zip(L, R):
    print(lval, rval)
````

Można połączyć dowolną liczbę obiektów iterowalnych, a jeśli mają one różną długość, to długość wyniku operacji ``zip`` będzie zależała od najkrótszego z nich.

### ``map`` i ``filter``
Iterator ``map`` pobiera funkcję i stosuje ją do wartości zawartych w iteratorze:

````python
# find the first 10 square numbers
square = lambda x: x ** 2
for val in map(square, range(10)):
    print(val, end=' ')
````

Iterator ``filter`` wygląda podobnie, z tą różnicą, że przekazuje tylko te wartości, dla których funkcja filtrująca zwraca wartość True:

````python
# find values up to 10 for which x % 2 is zero
is_even = lambda x: x % 2 == 0
for val in filter(is_even, range(10)):
    print(val, end=' ')
````

Funkcje `map` i `filter`, wraz z funkcją `reduce` (znajdującą się w module `functools` języka Python), stanowią podstawowe elementy stylu *programowania funkcjonalnego*, który – choć nie jest dominującym stylem programowania w świecie Pythona – ma swoich zagorzałych zwolenników (zobacz na przykład bibliotekę [pytoolz](https://toolz.readthedocs.org/en/latest/)).

### Iteratory jako argumenty funkcji

W ``*args`` i ``**kwargs`` widzieliśmy, że ``*args`` i ``**kwargs`` można wykorzystać do przekazywania sekwencji i słowników do funkcji.
Okazuje się, że składnia ``*args`` działa nie tylko z sekwencjami, ale z dowolnym iteratorem:

````python
print(*range(10))
````

Na przykład możemy trochę się pobawić i skompresować poprzedni przykład z `map` do następującej postaci:

````python
print(*map(lambda x: x ** 2, range(10)))
````

Dzięki tej sztuczce możemy odpowiedzieć na odwieczne pytanie pojawiające się na forach dla osób uczących się języka Python: dlaczego nie ma funkcji `unzip()`, która działałaby odwrotnie niż `zip()`?
Jeśli zamkniesz się w ciemnej szafie i pomyślisz o tym przez chwilę, może dojdziesz do wniosku, że przeciwieństwem ``zip()`` jest... ``zip()``! Kluczem jest to, że ``zip()`` może połączyć dowolną liczbę iteratorów lub sekwencji. Zobacz:

````python
L1 = (1, 2, 3, 4)
L2 = ('a', 'b', 'c', 'd')
````

````python
z = zip(L1, L2)
print(*z)
````

````python
z = zip(L1, L2)
new_L1, new_L2 = zip(*z)
print(new_L1, new_L2)
````

## Specjalistyczne iteratory: ``itertools``

Krótko omówiliśmy nieskończony iterator ``range`` oraz funkcję ``itertools.count``.
Moduł ``itertools`` zawiera całą gamę przydatnych iteratorów; warto zapoznać się z tym modułem, aby sprawdzić, jakie funkcje są w nim dostępne.
Jako przykład rozważmy funkcję ``itertools.permutations``, która iteruje nad wszystkimi permutacjami sekwencji:

````python
from itertools import permutations
p = permutations(range(3))
print(*p)
````

Podobnie funkcja `itertools.combinations` przegląda wszystkie unikalne kombinacje `N` wartości z listy:

````python
from itertools import combinations
c = combinations(range(4), 2)
print(*c)
````

Nieco powiązany z tym jest iterator `product`, który przechodzi po wszystkich parach elementów występujących między dwoma lub więcej obiektami iterowalnymi:

````python
from itertools import product
p = product('ab', range(3))
print(*p)
````

# Rozumienie list

Jeśli przeczytasz wystarczająco dużo kodu w Pythonie, w końcu natkniesz się na zwięzłą i wydajną konstrukcję znaną jako *list comprehension*.
To jedna z funkcji Pythona, w której z pewnością się zakochasz, jeśli jeszcze jej nie używałeś; wygląda mniej więcej tak:

````python
[i for i in range(20) if i % 3 > 0]
````

W rezultacie otrzymujemy listę liczb, z której wykluczono wielokrotności liczby 3.
Choć ten przykład może początkowo wydawać się nieco zagmatwany, wraz z rosnącą znajomością języka Python czytanie i pisanie wyrażenia listowego stanie się dla Ciebie czymś naturalnym.

## Podstawowe wyrażenia listowe
Wyrażenia listowe to po prostu sposób na skrócenie pętli for służącej do tworzenia listy do jednego krótkiego i czytelnego wiersza.
Oto przykład pętli, która tworzy listę pierwszych 12 liczb całkowitych podniesionych do kwadratu:

````python
L = []
for n in range(12):
    L.append(n ** 2)
L
````

Odpowiednikiem tego wyrażenia w postaci listy jest następujący zapis:

````python
[n ** 2 for n in range(12)]
````

Podobnie jak w przypadku wielu instrukcji w języku Python, znaczenie tej instrukcji można niemal odczytać w zwykłym języku polskim: „utwórz listę składającą się z kwadratów liczby ``n`` dla każdej wartości ``n`` do 12”.

Ta podstawowa składnia to zatem ``[``*``wyrażenie``* ``for`` *``zmienna``* ``in`` *``obiekt iterowalny``*``]``, gdzie *``wyrażenie``* jest dowolnym poprawnym wyrażeniem, *``zmienna``* jest nazwą zmiennej, a *``obiekt iterowalny``* jest dowolnym obiektem iterowalnym w Pythonie.

## Wielokrotne iteracje
Czasami chcesz utworzyć listę nie tylko na podstawie jednej wartości, ale dwóch. Aby to zrobić, po prostu dodaj kolejne wyrażenie ``for`` w wyrażeniu skróconym:

````python
[(i, j) for i in range(2) for j in range(3)]
````

Warto zauważyć, że drugie wyrażenie ``for`` pełni rolę indeksu wewnętrznego, zmieniając się najszybciej w powstałej liście.
Tego typu konstrukcję można rozszerzyć na trzy, cztery lub więcej iteratorów w ramach wyrażenia zrozumienia, choć w pewnym momencie ucierpi na tym czytelność kodu!

## Warunki w iteratorze
Można jeszcze bardziej kontrolować przebieg iteracji, dodając warunek na końcu wyrażenia.
W pierwszym przykładzie z tej sekcji iterowaliśmy po wszystkich liczbach od 1 do 20, pomijając jednak wielokrotności liczby 3.
Przyjrzyj się temu jeszcze raz i zwróć uwagę na konstrukcję:

````python
[val for val in range(20) if val % 3 > 0]
````

Wyrażenie ``(i % 3 > 0)`` zwraca wartość ``True``, chyba że ``val`` jest podzielne przez 3.
Również w tym przypadku znaczenie można od razu odczytać z języka polskiego: „Utwórz listę wartości dla każdej wartości do 20, ale tylko wtedy, gdy wartość ta nie jest podzielna przez 3”.
Gdy już się z tym oswoisz, jest to znacznie łatwiejsze do napisania – i do zrozumienia na pierwszy rzut oka – niż równoważna składnia pętli:

````python
L = []
for val in range(20):
    if val % 3:
        L.append(val)
L
````

## Warunki oparte na wartości
Jeśli masz doświadczenie w programowaniu w języku C, być może znasz jednowierszowe wyrażenia warunkowe z wykorzystaniem operatora  ``? ``:
``` C
int absval = (val < 0) ? -val : val
```
W języku Python istnieje bardzo podobna konstrukcja, z której najczęściej korzysta się w wyrażeniach listowych, funkcjach  ``lambda `` oraz w innych miejscach, gdzie pożądane jest proste wyrażenie:

````python
val = -10
val if val >= 0 else -val
````

Widzimy, że jest to po prostu powielenie działania wbudowanej funkcji ``abs()``, ale taka konstrukcja pozwala na realizację naprawdę ciekawych rozwiązań w ramach wyrażenia listowego.
Sprawa zaczyna się teraz nieco komplikować, ale można by zrobić coś takiego:

````python
[val if val % 2 else -val
 for val in range(20) if val % 3]
````

Zwróć uwagę na znak końca linii w wyrażeniu listowym przed wyrażeniem ``for``: jest to poprawne w języku Python i często stanowi dobry sposób na podzielenie długich wyrażeń listowych w celu zwiększenia czytelności.
Przyjrzyj się temu: tworzymy listę, pomijając wielokrotności liczby 3 i odwracając znak wszystkich wielokrotności liczby 2.

Kiedy już zrozumiesz dynamikę list comprehensions, przejście do innych typów comprehensions jest proste. Składnia jest w dużej mierze taka sama; jedyną różnicą jest rodzaj nawiasów, których używasz.

Na przykład, za pomocą nawiasów klamrowych możesz utworzyć ``set`` za pomocą *set comprehension*:

````python
{n**2 for n in range(12)}
````

Przypomnijmy, że ``set`` to zbiór, który nie zawiera elementów powtarzających się.
Wyrażenie definiujące zbiór uwzględnia tę zasadę i usuwa wszelkie powtarzające się elementy:

````python
{a % 3 for a in range(1000)}
````

Wystarczy niewielka zmiana – wystarczy dodać dwukropek (``:``), aby utworzyć *wyrażenie słownikowe*:

````python
{n:n**2 for n in range(6)}
````

Wreszcie, jeśli zamiast nawiasów kwadratowych użyjesz nawiasów okrągłych, otrzymasz tak zwane *wyrażenie generujące*:

````python
(n**2 for n in range(12))
````

Wyrażenie generujące to w zasadzie lista zrozumiała, w której elementy są generowane w miarę potrzeb, a nie wszystkie naraz.

# Moduły i pakiety

Jedną z cech języka Python, która sprawia, że nadaje się on do szerokiego zakresu zadań, jest fakt, że jest on dostarczany z „wbudowanymi funkcjami” – innymi słowy, standardowa biblioteka Pythona zawiera przydatne narzędzia do realizacji różnorodnych zadań.
Ponadto istnieje rozbudowany ekosystem narzędzi i pakietów innych producentów, które oferują bardziej specjalistyczne funkcje.
W tym miejscu przyjrzymy się importowaniu modułów biblioteki standardowej, narzędziom do instalowania modułów innych firm oraz opisowi tego, jak można tworzyć własne moduły.

## Ładowanie modułów: instrukcja ``import``

Do ładowania modułów wbudowanych i modułów innych producentów język Python udostępnia instrukcję ``import``.
Istnieje kilka sposobów korzystania z tej instrukcji, które pokrótce omówimy poniżej, zaczynając od najbardziej zalecanych, a kończąc na najmniej zalecanych.

### Wyraźne importowanie modułów

Wyraźne importowanie modułu pozwala zachować zawartość modułu w przestrzeni nazw.
Przestrzeń nazw służy następnie do odwoływania się do zawartości modułu poprzez umieszczenie między nimi znaku „``.``”.
Na przykład poniżej zaimportujemy wbudowany moduł ``math`` i obliczymy cosinus liczby pi:

````python
import math
math.cos(math.pi)
````

### Wyraźne importowanie modułów za pomocą aliasów

W przypadku dłuższych nazw modułów nie jest wygodne używanie pełnej nazwy modułu przy każdym dostępie do jakiegoś elementu.
Z tego powodu często używamy wzorca „``import ... as ...``”, aby utworzyć krótszy alias dla przestrzeni nazw.
Na przykład pakiet NumPy (Numerical Python), popularny pakiet zewnętrzny przydatny w nauce o danych, jest zgodnie z konwencją importowany pod aliasem ``np``:

````python
import numpy as np
np.cos(np.pi)
````

### Wyraźne importowanie zawartości modułu

Czasami zamiast importować przestrzeń nazw modułu, wystarczy zaimportować tylko kilka konkretnych elementów z tego modułu.
Można to zrobić za pomocą konstrukcji „``from ... import ...``”.
Na przykład możemy zaimportować tylko funkcję ``cos`` i stałą ``pi`` z modułu ``math``:

````python
from math import cos, pi
cos(pi)
````

### Niejawne importowanie zawartości modułu

Wreszcie, czasami przydatne jest zaimportowanie całej zawartości modułu do lokalnej przestrzeni nazw.
Można to zrobić za pomocą konstrukcji „``from ... import *``”:

````python
from math import *
sin(pi) ** 2 + cos(pi) ** 2
````

Z tego wzorca należy korzystać z umiarem, o ile w ogóle.
Problem polega na tym, że takie importy mogą czasami nadpisać nazwy funkcji, których nie zamierzało się nadpisywać, a niejawny charakter tej instrukcji utrudnia ustalenie, co uległo zmianie.

Na przykład Python posiada wbudowaną funkcję ``sum``, którą można wykorzystać do różnych operacji:

````python
help(sum)
````

Możemy to wykorzystać do obliczenia sumy ciągu, zaczynając od określonej wartości (w tym przypadku zaczniemy od ``-1``):

````python
sum(range(5), -1)
````

Zobaczmy teraz, co się stanie, jeśli wykonamy *dokładnie to samo wywołanie funkcji* po zaimportowaniu ``*`` z biblioteki ``numpy``:

````python
from numpy import *
````

````python
sum(range(5), -1)
````

Wynik różni się o jeden!
Powodem tego jest fakt, że instrukcja ``import *`` *zastępuje* wbudowaną funkcję ``sum`` funkcją ``numpy.sum``, która ma inną sygnaturę wywołania: w pierwszej z nich sumujemy ``range(5)`` zaczynając od ``-1``; w drugiej sumujemy ``range(5)`` wzdłuż ostatniej osi (oznaczonej przez ``-1``).
Jest to sytuacja, która może się zdarzyć, jeśli nie zachowamy ostrożności podczas używania „``import *``” – z tego powodu najlepiej tego unikać, chyba że dokładnie wiemy, co robimy.

````python
#GJ Supplement
import numpy as np

# Rows = Students (Alice, Bob, Charlie)
# Columns = Subjects (Math, Science, History)
scores = np.array([
    [80, 90, 100],  # Alice
    [70, 85, 90],   # Bob
    [60, 70, 80]    # Charlie
])
print("scores:", scores, end=f"\n")
print("shape:", shape(scores), " - 3 rows, 3 columns")

print("no axis", np.sum(scores), " - removing both dimensions, showing school's sum")

print("axis 0:", np.sum(scores, axis=0), " - removing rows, showing subject's sum")

print("axis 1:", np.sum(scores, axis=1), " - removing columns, showing student's sum")

print("axis -1:", np.sum(scores, axis=-1), " - last axis is related to columns")
````

## Prosta obróbka ciągów znaków w Pythonie

W przypadku podstawowej obróbki ciągów znaków wbudowane metody Pythona mogą okazać się niezwykle przydatne.
Jeśli masz doświadczenie w pracy z językiem C lub innym językiem niskiego poziomu, prostota metod Pythona z pewnością będzie dla Ciebie miłą odmianą.
Wcześniej omówiliśmy typ ciągu znaków w Pythonie oraz kilka z tych metod; teraz zagłębimy się w ten temat nieco bardziej

### Formatowanie ciągów znaków: zmiana wielkości liter

W języku Python zmiana wielkości liter w ciągu znaków jest dość prosta.
W tym miejscu przyjrzymy się metodom ``upper()``, ``lower()``, ``capitalize()``, ``title()`` oraz ``swapcase()``, posługując się jako przykładem poniższym nieuporządkowanym ciągiem znaków:

````python
fox = "tHe qUICk bROWn fOx."
````

Aby zamienić cały ciąg znaków na wielkie lub małe litery, można użyć odpowiednio metod ``upper()`` lub ``lower()``:

````python
fox.upper()
````

````python
fox.lower()
````

Często zdarza się, że trzeba wielką literą zapisać tylko pierwszą literę każdego słowa lub pierwszą literę każdego zdania.
Można to zrobić za pomocą metod ``title()`` i ``capitalize()``:

````python
fox.title()
````

````python
fox.capitalize()
````

Obudowy można zamienić za pomocą metody ``swapcase()``:

````python
fox.swapcase()
````

### Formatowanie ciągów znaków: dodawanie i usuwanie spacji

Często zdarza się też, że trzeba usunąć spacje (lub inne znaki) z początku lub końca ciągu znaków.
Podstawową metodą usuwania znaków jest metoda ``strip()``, która usuwa spacje z początku i końca ciągu:

````python
line = '         this is the content         '
line.strip()
````

Aby usunąć tylko spacje po prawej lub lewej stronie, użyj odpowiednio funkcji ``rstrip()`` lub ``lstrip()``:

````python
line.rstrip()
````

````python
line.lstrip()
````

Aby usunąć znaki inne niż spacje, można przekazać żądany znak do metody ``strip()``:

````python
num = "000000000000435"
num.strip('0')
````

Odwrotność tej operacji, czyli dodawanie spacji lub innych znaków, można wykonać za pomocą metod ``center()``, ``ljust()`` i ``rjust()``.

Na przykład za pomocą metody ``center()`` możemy wyśrodkować dany ciąg znaków w określonych odstępach:

````python
line = "this is the content"
line.center(30)
````

Similarly, ``ljust()`` and ``rjust()`` will left-justify or right-justify the string within spaces of a given length:

````python
line.ljust(30)
````

````python
line.rjust(30)
````

Wszystkie te metody akceptują dodatkowo dowolny znak, który zostanie użyty do wypełnienia luki.
Na przykład:

````python
'435'.rjust(10, '0')
````

Ponieważ uzupełnianie zerami jest tak powszechną potrzebą, Python udostępnia również metodę `zfill()`, która służy do uzupełniania ciągu znaków zerami od prawej strony:

````python
'435'.zfill(10)
````

### Wyszukiwanie i zamiana podciągów

Jeśli chcesz znaleźć wystąpienia określonego znaku w ciągu znaków, najlepszymi wbudowanymi metodami są ``find()``/``rfind()``, ``index()``/``rindex()`` oraz ``replace()``.

Metody ``find()`` i ``index()`` są bardzo podobne, ponieważ wyszukują pierwsze wystąpienie znaku lub podciągu w ciągu znaków i zwracają indeks podciągu:

````python
line = 'the quick brown fox jumped over a lazy dog'
line.find('fox')
````

````python
line.index('fox')
````

Jedyna różnica między funkcjami `find()` i `index()` polega na tym, jak zachowują się one w przypadku, gdy szukany ciąg znaków nie zostanie znaleziony; funkcja `find()` zwraca wartość `-1`, natomiast funkcja `index()` generuje błąd `ValueError`:

````python
line.find('bear')
````

````python
line.index('bear')
````

Powiązane funkcje ``rfind()`` i ``rindex()`` działają podobnie, z tą różnicą, że szukają pierwszego wystąpienia od końca, a nie od początku ciągu znaków:

````python
line.rfind('a')
````

W szczególnym przypadku sprawdzania, czy na początku lub na końcu ciągu znaków występuje podciąg, język Python udostępnia metody ``startswith()`` i ``endswith()``:

````python
line.endswith('dog')
````

````python
line.startswith('fox')
````

Aby pójść o krok dalej i zastąpić dany podciąg nowym ciągiem znaków, można skorzystać z metody ``replace()``.
W tym przypadku zastąpmy ``'brown'`` przez ``'red'``:

````python
line.replace('brown', 'red')
````

Funkcja ``replace()`` zwraca nowy ciąg znaków i zastępuje wszystkie wystąpienia podanego ciągu:

````python
line.replace('o', '--')
````

### Dzielenie i rozdzielanie ciągów znaków

Jeśli chcesz znaleźć podciąg, *a następnie* podzielić ciąg znaków w oparciu o jego położenie, metody ``partition()`` i/lub ``split()`` są tym, czego szukasz.
Obie zwrócą sekwencję podciągów.

Metoda ``partition()`` zwraca krotkę zawierającą trzy elementy: podciąg przed pierwszym wystąpieniem punktu podziału, sam punkt podziału oraz podciąg po nim:

````python
line.partition('fox')
````

Metoda ``rpartition()`` działa podobnie, ale przeszukuje ciąg znaków od prawej strony.

Metoda ``split()`` jest prawdopodobnie bardziej przydatna; wyszukuje *wszystkie* wystąpienia znaku rozdzielającego i zwraca podciągi znajdujące się pomiędzy nimi.
Domyślnie rozdziela ciąg na podstawie dowolnej spacji, zwracając listę poszczególnych słów w ciągu znaków:

````python
line.split()
````

Podobną metodą jest ``splitlines()``, która dzieli ciąg znaków na podstawie znaków nowej linii.
Zróbmy to na przykładzie haiku, powszechnie przypisywanym XVII-wiecznemu poecie Matsuo Bashō:

````python
haiku = """matsushima-ya
aah matsushima-ya
matsushima-ya"""

haiku.splitlines()
````

````python
#GJ Supplement
# "matsushima" to region w japonii, a samo haiku najprawdopodobniej nie jest autostwa Bashō, chociaż niemniej może zachwycać swoją prostotą.
# Poniżej jedno z ostatnich dzieł Bashō:
# "Chory w podróży śnię, że polem wymarłym biegam bez końca... - Matsuo Bashō"
````

Pamiętaj, że jeśli chcesz cofnąć działanie metody ``split()``, możesz użyć metody ``join()``, która zwraca ciąg znaków utworzony z punktu podziału i obiektu iterowalnego:

````python
'--'.join(['1', '2', '3'])
````

Często stosowanym sposobem jest użycie znaku specjalnego ``\n`` (nowa linia) w celu połączenia wcześniej rozdzielonych wierszy i odtworzenia danych wejściowych:

````python
print("\n".join(['matsushima-ya', 'aah matsushima-ya', 'matsushima-ya']))
````

## Ciągi formatujące

W poprzednich metodach nauczyliśmy się, jak wyodrębniać wartości z ciągów znaków oraz jak przekształcać same ciągi znaków do pożądanego formatu.
Innym zastosowaniem metod łańcuchowych jest manipulowanie *reprezentacjami* wartości innych typów w postaci ciągów znaków.
Oczywiście reprezentacje łańcuchowe można zawsze uzyskać za pomocą funkcji ``str()``; na przykład:

````python
pi = 3.14159
str(pi)
````

````python
"The value of pi is " + str(pi)
````

Bardziej elastycznym sposobem jest użycie *ciągów formatujących*, czyli ciągów znaków zawierających specjalne znaczniki (oznaczone nawiasami klamrowymi), w które wstawiane są wartości sformatowane jako ciągi znaków.
Oto prosty przykład:

````python
"The value of pi is {}".format(pi)
````

W obrębie znacznika ``{}`` można również podać informacje o tym, *co* dokładnie ma się tam pojawić.
Jeśli podasz liczbę, będzie ona odnosić się do indeksu argumentu, który ma zostać wstawiony:

````python
"""First letter: {0}. Last letter: {1}.""".format('A', 'Z')
````

Jeśli podasz ciąg znaków, będzie on odnosił się do klucza dowolnego argumentu słownego:

````python
"""First letter: {first}. Last letter: {last}.""".format(last='Z', first='A')
````

Wreszcie, w przypadku danych liczbowych można używać kodów formatowania, które określają sposób konwersji wartości na ciąg znaków.
Na przykład, aby wyświetlić liczbę jako liczbę zmiennoprzecinkową z trzema cyframi po przecinku, można użyć następującego zapisu:

````python
"pi = {0:.3f}".format(pi)
````

Tak jak poprzednio, symbol „``0``” odnosi się tutaj do indeksu wartości, która ma zostać wstawiona.
Znak „``:``” oznacza, że po nim następują kody formatujące.
Wyrażenie „``.3f``” określa pożądaną precyzję: trzy cyfry po przecinku w formacie zmiennoprzecinkowym.

Ten sposób określania formatu jest bardzo elastyczny, a przedstawione tu przykłady to zaledwie niewielka część dostępnych opcji formatowania.
