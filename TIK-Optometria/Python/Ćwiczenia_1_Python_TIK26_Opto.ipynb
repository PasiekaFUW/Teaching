# Szybkie zapoznanie ze składnią Pythona

Język Python został pierwotnie opracowany jako język dydaktyczny, ale dzięki swojej łatwości użytkowania i przejrzystej składni zyskał popularność zarówno wśród początkujących, jak i ekspertów.
Przejrzystość składni Pythona sprawiła, że niektórzy nazywają go „wykonywalnym pseudokodem" i rzeczywiście często znacznie łatwiej jest przeczytać i zrozumieć skrypt w Pythonie niż podobny skrypt napisany np. w języku C.
W tym miejscu zaczniemy omawiać główne cechy składni Pythona.

Składnia odnosi się do struktury języka (tj. tego, co składa się na poprawnie sformułowany program).
Na razie nie będziemy skupiać się na semantyce – znaczeniu słów i symboli w składni – ale wrócimy do tego w późniejszym czasie.

Rozważmy następujący przykład kodu:

````python
# set the midpoint
midpoint = 5

# make two empty lists
lower = []; upper = []

# split the numbers into lower and upper
for i in range(10):
    if (i < midpoint):
        lower.append(i)
    else:
        upper.append(i)

print("lower:", lower)
print("upper:", upper)
````

Ten skrypt jest nieco zabawny, ale w zwięzły sposób ilustruje kilka ważnych aspektów składni języka Python.
Przejrzyjmy go i omówmy niektóre cechy składniowe języka Python

## Komentarze są oznaczane znakiem „#”
Skrypt zaczyna się od komentarza:
``` python
# set the midpoint
```
Komentarze w języku Python są oznaczone znakiem hash (``#``), a wszystko, co znajduje się w wierszu następującym po tym znaku, jest ignorowane przez interpreter.
Oznacza to na przykład, że można stosować komentarze samodzielne, takie jak ten właśnie pokazany, a także komentarze wbudowane, które następują po instrukcji. Na przykład:
``` python
x += 2  # skrót dla x = x + 2
```
Python nie posiada składni dla komentarzy wielowierszowych, takich jak składnia ``/* ... */`` używana w językach C i C++, choć ciągi wielowierszowe są często używane jako zamiennik komentarzy wielowierszowych (więcej na ten temat w późniejszych notebookach).



## Koniec linii oznacza koniec instrukcji
Kolejna linia w skrypcie wygląda następująco:
``` python
midpoint = 5
```
Jest to operacja przypisania, w której utworzyliśmy zmienną o nazwie „midpoint” i przypisaliśmy jej wartość „5”.
Zauważ, że koniec tej instrukcji jest po prostu oznaczony przez koniec linii.
Różni się to od języków takich jak C i C++, gdzie każda instrukcja musi kończyć się średnikiem (``;``).

W Pythonie, jeśli chcesz, aby instrukcja była kontynuowana w następnej linii, możesz użyć znaku „``\``”, aby to zaznaczyć:

````python
x = 1 + 2 + 3 + 4 +\
    5 + 6 + 7 + 8
````

Ale da się to zrobić również bez tego znaku:

````python
x = (1 + 2 + 3 + 4 +
     5 + 6 + 7 + 8)
````

Ta druga wersja jest bardziej popularna i zalecana.

## Średnik może opcjonalnie kończyć instrukcję
Czasami warto umieścić kilka instrukcji w jednym wierszu.
Kolejna część skryptu wygląda następująco:
``` python
lower = []; upper = []
```
Pokazuje to przykład tego, jak średnik (``;``), znany z języka C, może być opcjonalnie używany w Pythonie do umieszczenia dwóch instrukcji w jednym wierszu.
Pod względem funkcjonalnym jest to całkowicie równoważne z zapisaniem
``` python
lower = []
upper = []
```
W większości przewodników stylistycznych dotyczących języka Python odradza się używanie średnika do umieszczania wielu instrukcji w jednym wierszu, choć czasami okazuje się to wygodne.

## Wcięcia: spacje mają znaczenie!
Przejdźmy teraz do głównego bloku kodu:
``` Python
for i in range(10):
    if i < midpoint:
        lower.append(i)
    else:
        upper.append(i)
```
Jest to złożona instrukcja sterująca przebiegiem programu, zawierająca pętlę i warunek – za chwilę przyjrzymy się bliżej tego typu instrukcjom.
Na razie weźmy pod uwagę, że ilustruje to prawdopodobnie najbardziej kontrowersyjną cechę składni Pythona: spacje mają znaczenie!

W językach programowania *blok* kodu to zbiór instrukcji, które powinny być traktowane jako całość.
Na przykład w języku C bloki kodu są oznaczone nawiasami klamrowymi:
``` C
// Kod w języku C
for(int i=0; i<100; i++)
   {
      // nawiasy klamrowe wskazują blok kodu
      total += i;
   }
```
W Pythonie bloki kodu są oznaczone *wcięciem*:
``` python
for i in range(100):
    # wcięcie wskazuje blok kodu
    total += i
```
W Pythonie bloki kodu z wcięciem są zawsze poprzedzone dwukropkiem (``:``) w poprzednim wierszu.

Stosowanie wcięć pomaga zachować jednolity, czytelny styl, który wielu osobom wydaje się atrakcyjny w kodzie Pythona.
Może to jednak być mylące dla osób niezaznajomionych z tym językiem; na przykład poniższe dwa fragmenty kodu dadzą różne wyniki:
```python
>>> if x < 4:         >>> if x < 4:
...     y = x * 2     ...     y = x * 2
...     print(x)      ... print(x)
```
W fragmencie po lewej stronie `print(x)` znajduje się w wciętym bloku i zostanie wykonane tylko wtedy, gdy `x` jest mniejsze od `4`.
W fragmencie po prawej stronie `print(x)` znajduje się poza blokiem i zostanie wykonane niezależnie od wartości `x`!

Sposób, w jaki Python wykorzystuje znaczące spacje, często zaskakuje programistów przyzwyczajonych do innych języków, ale w praktyce może to prowadzić do znacznie bardziej spójnego i czytelnego kodu niż w językach, które nie wymuszają wcięć bloków kodu.
Jeśli sposób wykorzystania spacji w Pythonie nie przypadł Ci do gustu, zachęcam Cię do spróbowania: być może z czasem zaczniesz to doceniać.

Na koniec należy pamiętać, że *ilość* spacji używanych do wcięć bloków kodu zależy od użytkownika, o ile jest ona spójna w całym skrypcie.
Zgodnie z konwencją większość przewodników stylistycznych zaleca wcięcia bloków kodu o cztery spacje i właśnie tej konwencji będziemy przestrzegać w niniejszym notebooku.
Należy zauważyć, że wiele edytorów tekstu, takich jak Emacs i Vim, zawiera tryby Pythona, które automatycznie stosują wcięcia o szerokości czterech spacji.

## Spacje *wewnątrz* wierszy nie mają znaczenia
Chociaż zasada *sensownego stosowania spacji* ma zastosowanie do spacji *przed* wierszami (które wyznaczają blok kodu), spacje *wewnątrz* wierszy kodu w języku Python nie mają znaczenia.
Na przykład wszystkie trzy poniższe wyrażenia są równoważne:

````python
x=1+2; print(x)
y = 1 + 2; print(y)
z             =        1    +                2; print(z)
````

Nadużywanie tej elastyczności może prowadzić do problemów z czytelnością kodu – w rzeczywistości nadużywanie spacji jest często jednym z głównych sposobów celowego zaciemniania kodu (co niektórzy robią dla zabawy).
Efektywne wykorzystanie spacji może sprawić, że kod będzie znacznie bardziej czytelny,
zwłaszcza w przypadkach, gdy operatory występują jeden po drugim – porównajmy dwa poniższe wyrażenia służące do potęgowania przez liczbę ujemną:
``` python
x=10**-2
```
z
``` python
x = 10 ** -2
```
Uważam, że druga wersja z odstępami jest znacznie bardziej czytelna na pierwszy rzut oka.
Większość przewodników stylistycznych dotyczących języka Python zaleca stosowanie pojedynczego odstępu wokół operatorów dwuargumentowych oraz brak odstępów wokół operatorów jednoargumentowych.
Operatorami w języku Python zajmiemy się bardziej szczegółowo później.

## Nawiasy służą do grupowania lub wywoływania

W powyższym fragmencie kodu widzimy dwa zastosowania nawiasów.
Po pierwsze, można ich używać w typowy sposób do grupowania instrukcji lub operacji matematycznych:

````python
2 * (3 + 4)
````

Można ich również używać do zaznaczenia, że wywoływana jest *funkcja*.
W poniższym fragmencie kodu funkcja ``print()`` służy do wyświetlenia zawartości zmiennej (zobacz ramka).
Wywołanie funkcji jest oznaczone parą nawiasów otwierających i zamykających, w których znajdują się *argumenty* funkcji:

````python
print('first value:', 1)
print('second value:', 2)
````

Niektóre funkcje można wywołać bez żadnych argumentów, jednak nawet w takim przypadku należy użyć nawiasów otwierających i zamykających, aby zaznaczyć, że funkcja została wywołana.
Przykładem jest metoda „sort” stosowana do list:

````python
L = [4,2,3,1]
L.sort()
print(L)
````

Znak „``()``” po słowie „sort” oznacza, że funkcja powinna zostać wykonana, i jest wymagany nawet wtedy, gdy nie są potrzebne żadne argumenty.

## Na marginesie: Uwaga dotycząca funkcji ``print()``

Powyżej posłużyliśmy się przykładem funkcji ``print()``.
Funkcja ``print()`` jest jednym z elementów, które uległy zmianie między wersjami Pythona *2.x* a *3.x*. W Pythonie 2 `print` działało jak instrukcja: można było napisać

# Tylko w Pythonie 2!

Jest to jedna z wielu konstrukcji, które nie są kompatybilne wstecznie między Pythonem 2 a 3.
Spotyka się przykłady napisane w obu wersjach Pythona, a obecność instrukcji „print” zamiast funkcji „print()” jest często jedną z pierwszych oznak, że masz do czynienia z kodem w Pythonie 2.

## Zmienne w Pythonie to wskaźniki

Przypisywanie wartości zmiennym w Pythonie jest tak proste, jak umieszczenie nazwy zmiennej po lewej stronie znaku równości (``=``):

```python
# przypisz 4 do zmiennej x
x = 4
```

Może się to wydawać oczywiste, ale jeśli masz błędny model mentalny tego, jak działa ta operacja, sposób działania Pythona może wydawać się zagmatwany.
Krótko omówimy to tutaj.

W wielu językach programowania zmienne najlepiej postrzegać jako pojemniki lub zbiorniki, do których umieszcza się dane.
Na przykład w języku C, gdy piszesz

```C
// Kod w języku C
int x = 4;
```

w zasadzie definiujesz „zbiornik pamięci” o nazwie ``x`` i umieszczasz w nim wartość ``4``.
Natomiast w Pythonie zmienne najlepiej traktować nie jako pojemniki, ale jako wskaźniki.
W Pythonie więc, gdy piszesz

```python
x = 4
```

w zasadzie definiujesz *wskaźnik* o nazwie ``x``, który wskazuje na jakiś inny pojemnik zawierający wartość ``4``.
Zwróć uwagę na jedną z konsekwencji tego podejścia: ponieważ zmienne w Pythonie po prostu wskazują na różne obiekty, nie ma potrzeby „deklarowania” zmiennej, ani nawet wymagania, aby zmienna zawsze wskazywała na informacje tego samego typu!
W tym sensie mówi się, że Python jest *językiem dynamicznie typowanym*: nazwy zmiennych mogą wskazywać na obiekty dowolnego typu.
W Pythonie możesz więc robić takie rzeczy:

````python
x = 1         # x jest liczbą całkowitą (integer)
x = 'hello'   # teraz x jest ciągiem znaków (string)
x = [1, 2, 3] # teraz x jest listą (list)
````

Chociaż użytkownicy języków statycznie typowanych mogą tęsknić za bezpieczeństwem typów, jakie zapewniają deklaracje takie jak te w języku C,

```C
int x = 4;
```

to właśnie to dynamiczne typowanie jest jednym z elementów, które sprawiają, że Python jest tak szybki w pisaniu i łatwy do czytania.

Istnieje jednak konsekwencja tego podejścia „zmienna jako wskaźnik”, o której należy pamiętać.
Jeśli mamy dwie nazwy zmiennych wskazujące na ten sam *zmienny* obiekt, to zmiana jednej z nich spowoduje zmianę drugiej!
Na przykład, stwórzmy i zmodyfikujmy listę:

````python
x = [1, 2, 3]
y = x
````

Utworzyliśmy dwie zmienne „x” i „y”, które wskazują na ten sam obiekt.
Z tego powodu, jeśli zmodyfikujemy listę za pomocą jednej z jej nazw, zauważymy, że „druga” lista również ulegnie zmianie:

````python
print(y)
x.append(4) # dodaj 4 do listy, na którą wskazuje x
print(y) # lista y również się zmieniła!
````

Takie zachowanie może wydawać się niezrozumiałe, jeśli błędnie postrzegasz zmienne jako pojemniki zawierające dane.
Jeśli jednak prawidłowo traktujesz zmienne jako wskaźniki do obiektów, to zachowanie to nabiera sensu.

Zauważ również, że jeśli użyjemy znaku „``=``” do przypisania innej wartości do zmiennej ``x``, nie wpłynie to na wartość zmiennej ``y`` – przypisanie polega po prostu na zmianie obiektu, na który wskazuje zmienna:

````python
x = 'coś innego'
print(y)  # y nie ulega zmianie
````

Znowu ma to sens, jeśli potraktujemy „x” i „y” jako wskaźniki, a operator „=” jako operację zmieniającą adres, na który wskazuje dana nazwa.

Można by się zastanawiać, czy ta koncepcja wskaźników nie utrudnia śledzenia operacji arytmetycznych w Pythonie, ale Python jest skonstruowany tak, że nie stanowi to problemu. Liczby, ciągi znaków i inne *typy proste* są niezmienne: nie można zmienić ich wartości – można jedynie zmienić to, na jakie wartości wskazują zmienne.
Dlatego na przykład całkowicie bezpieczne jest wykonywanie operacji takich jak poniższe:

````python
x = 10
y = x
x += 5  # zwiększ x o 5
print("x =", x)
print("y =", y)
````

Kiedy wywołujemy wyrażenie „x += 5”, nie modyfikujemy wartości obiektu „10”, na który wskazuje zmienna „x”; zmieniamy raczej samą zmienną „x” tak, aby wskazywała na nowy obiekt typu integer o wartości „15”.
Z tego powodu operacja ta nie ma wpływu na wartość zmiennej „y”.

## Wszystko jest obiektem

Python jest językiem programowania obiektowego i w Pythonie wszystko jest obiektem.

Wyjaśnijmy, co to oznacza. Wcześniej widzieliśmy, że zmienne są po prostu wskaźnikami, a same nazwy zmiennych nie zawierają informacji o typie.
To skłania niektórych do błędnego twierdzenia, że Python jest językiem beztypowym. Ale tak nie jest!
Rozważmy następujący przykład:

````python
x = 4
print(type(x))
x = 'hello'
print(type(x))
x = 3.14159
print(type(x))
````

W języku Python istnieją typy danych; jednakże typy te nie są powiązane z nazwami zmiennych, lecz *z samymi obiektami*.

W językach programowania obiektowego, takich jak Python, *obiekt* to jednostka zawierająca dane wraz z powiązanymi metadanymi i/lub funkcjonalnością.
W Pythonie wszystko jest obiektem, co oznacza, że każda jednostka posiada pewne metadane (zwane *atrybutami*) oraz powiązaną funkcjonalność (zwaną *metodami*).
Dostęp do tych atrybutów i metod uzyskuje się za pomocą składni kropkowej.

Na przykład, wcześniej widzieliśmy, że listy posiadają metodę ``append``, która dodaje element do listy, a dostęp do niej uzyskuje się za pomocą składni kropkowej („``.``”):

````python
L = [1, 2, 3]
L.append(100)
print(L)
````

Chociaż można by się spodziewać, że obiekty złożone, takie jak listy, posiadają atrybuty i metody, to czasem zaskakujące jest to, że w języku Python nawet proste typy mają przypisane atrybuty i metody.
Na przykład typy liczbowe posiadają atrybuty „real” i „imag”, które zwracają część rzeczywistą i urojoną wartości, jeśli traktuje się ją jako liczbę zespoloną:

````python
x = 4.5
print(x.real, "+", x.imag, 'i')
````

Metody są podobne do atrybutów, z tą różnicą, że są to funkcje, które można wywoływać za pomocą nawiasów otwierających i zamykających.
Na przykład liczby zmiennoprzecinkowe mają metodę o nazwie „is_integer”, która sprawdza, czy dana wartość jest liczbą całkowitą:

````python
x = 4.5
print(x.is_integer())
x = 4.0
print(x.is_integer())
````

Kiedy mówimy, że w Pythonie wszystko jest obiektem, mamy na myśli, że naprawdę *wszystko* jest obiektem – nawet atrybuty i metody obiektów są same w sobie obiektami posiadającymi własne informacje o „typie”:

````python
type(x.is_integer)
````

W poprzednim rozdziale zaczęliśmy przyglądać się semantyce zmiennych i obiektów w języku Python; w tym rozdziale zagłębimy się w semantykę różnych *operatorów* dostępnych w tym języku.

## Operacje arytmetyczne
W języku Python zaimplementowano siedem podstawowych operatorów arytmetycznych, z których dwa mogą pełnić również funkcję operatorów jednoargumentowych.
Zostały one zestawione w poniższej tabeli:

| Operator     | Name           | Description                                            |
|--------------|----------------|--------------------------------------------------------|
| ``a + b``    | Addition       | Suma ``a`` i ``b``                                 |
| ``a - b``    | Subtraction    | Różnica ``a`` i ``b``                          |
| ``a * b``    | Multiplication | Iloczyn ``a`` i ``b``                             |
| ``a / b``    | True division  | Iloraz ``a`` i ``b``                            |
| ``a // b``   | Floor division | Iloraz ``a`` i ``b``, bez części dziesiętnych  |
| ``a % b``    | Modulus        | Całkowita reszta z dzielenia ``a`` przez ``b``     |
| ``a ** b``   | Exponentiation | ``a`` podniesione do potęgi ``b``                     |
| ``-a``       | Negation       | Liczba przeciwna do ``a``                                  |
| ``+a``       | Unary plus     | ``a`` bez zmian (rzadko używana operacja)                          |

Operatory te można stosować i łączyć w intuicyjny sposób, używając standardowych nawiasów do grupowania operacji.
Na przykład:

````python
# addition, subtraction, multiplication
print((4 + 8) * (6.5 - 3))

# True division
print(11 / 2)

# Floor division
print(11 // 2)
````

Operator dzielenia z obcięciem został wprowadzony w Pythonie 3; pracując w Pythonie 2, należy pamiętać, że standardowy operator dzielenia (``/``) działa jak dzielenie z obcięciem w przypadku liczb całkowitych, a jak zwykłe dzielenie w przypadku liczb zmiennoprzecinkowych.

Na koniec wspomnę o ósmym operatorze arytmetycznym, który został dodany w Pythonie 3.5: operatorze ``a @ b``, który ma na celu wskazanie *iloczynu macierzowego* ``a`` i ``b``, do wykorzystania w różnych pakietach algebry liniowej.

## Operacje bitowe
Oprócz standardowych operacji arytmetycznych język Python udostępnia operatory służące do wykonywania operacji logicznych na bitach liczb całkowitych.
Są one znacznie rzadziej używane niż standardowe operacje arytmetyczne, ale warto wiedzieć o ich istnieniu.
Sześć operatorów bitowych zestawiono w poniższej tabeli:

| Operator     | Nazwa            | Opis                                 |
|--------------|-----------------|---------------------------------------------|
| ``a & b``    | Bitowe AND     | Bity zdefiniowane zarówno w ``a``, jak i w ``b``        |
| <code>a &#124; b</code>| Bitowe OR      | Bity zdefiniowane w ``a`` lub w ``b`` lub w obu      |
| ``a ^ b``    | Bitowe XOR     | Bity zdefiniowane w ``a`` lub ``b``, ale nie w obu |
| ``a << b``   | Przesunięcie bitów w lewo  | Przesunięcie bitów ``a`` w lewo o ``b`` jednostek     |
| ``a >> b``   | Przesunięcie bitów w prawo | Przesunięcie bitów ``a`` w prawo o ``b`` jednostek    |
| „~a”       | Operacja NOT bitowa     | Negacja bitowa „a”                          |

Te operatory bitowe mają sens tylko w kontekście binarnej reprezentacji liczb, którą można zobaczyć za pomocą wbudowanej funkcji „bin”:

````python
bin(10)
````

Wynik ma przedrostek '0b', co oznacza zapis binarny.
Pozostałe cyfry wskazują, że liczba 10 jest wyrażona jako suma $1 \cdot 2^3 + 0 \cdot 2^2 + 1 \cdot 2^1 + 0 \cdot 2^0$.
Podobnie możemy zapisać:

````python
bin(4)
````

Teraz, korzystając z operacji logicznej OR, możemy znaleźć liczbę, która łączy bity z pozycji 4 i 10:

````python
print(4 | 10)
print(bin(4 | 10))
````

Te operatory bitowe nie są tak bezpośrednio przydatne jak standardowe operatory arytmetyczne, ale warto się z nimi zapoznać przynajmniej raz, aby zrozumieć, jaką kategorię operacji wykonują.
W szczególności użytkownicy znający inne języki programowania mają czasem skłonność do używania operatora XOR (tj. „a ^ b”), gdy w rzeczywistości mają na myśli potęgowanie (tj. „a ** b”).

W przypadku obiektów zmiennych, takich jak listy, tablice lub ramki danych, te rozszerzone operacje przypisania różnią się nieco od swoich bardziej rozbudowanych odpowiedników: modyfikują one zawartość oryginalnego obiektu, zamiast tworzyć nowy obiekt do przechowywania wyniku.

Jak można się spodziewać, operatory te są wyrażane za pomocą słów „and”, „or” oraz „not”:

````python
x = 4
(x < 6) and (x > 2)
````

Miłośnicy algebry Boole'a mogą zauważyć, że nie uwzględniono operatora XOR; można go oczywiście uzyskać na kilka sposobów, łącząc inne operatory.
Poza tym, sprytnym trikiem, którego można użyć do wykonania operacji XOR na wartościach logicznych, jest następująca metoda:

````python
# (x > 1) xor (x < 10)
(x > 1) != (x < 10)
````

Tego rodzaju operacje boolowskie okażą się niezwykle przydatne, gdy zaczniemy omawiać *instrukcje sterujące przebiegiem programu*, takie jak instrukcje warunkowe i pętle.

Czasami pewną trudnością w tym języku jest rozróżnienie, kiedy należy stosować operatory boolowskie (``and``, ``or``, ``not``), a kiedy operacje bitowe (``&``, ``|``, ``~``).
Odpowiedź tkwi w ich nazwach: operatorów boolowskich należy używać, gdy chcesz obliczyć *wartości boolowskie (tj. prawdziwość lub fałsz) całych instrukcji*.
Operacji bitowych należy używać, gdy chcesz *działać na poszczególnych bitach lub składnikach danych obiektów*.

Jak widzieliśmy w poprzedniej sekcji, zmienne w Pythonie są wskaźnikami. Operator „is” sprawdza, czy dwie zmienne wskazują na ten sam kontener (obiekt), a nie odnosi się do zawartości tego kontenera.
Mając to na uwadze, w większości przypadków, gdy początkujący programista ma ochotę użyć „is”, tak naprawdę ma na myśli „==”.

### Na marginesie: Dokładność liczb zmiennoprzecinkowych
W przypadku arytmetyki zmiennoprzecinkowej należy pamiętać, że jej dokładność jest ograniczona, co może powodować niestabilność testów równości. Na przykład:

````python
0.1 + 0.2 == 0.3
````

Dlaczego tak się dzieje? Okazuje się, że nie jest to zachowanie charakterystyczne wyłącznie dla języka Python, lecz wynika z formatu o stałej precyzji, w jakim przechowywane są liczby zmiennoprzecinkowe w postaci binarnej – formatu stosowanego przez większość, jeśli nie wszystkie, platform obliczeniowych przeznaczonych do zastosowań naukowych.
Wszystkie języki programowania wykorzystujące liczby zmiennoprzecinkowe przechowują je w stałej liczbie bitów, co powoduje, że niektóre liczby są reprezentowane jedynie w przybliżeniu.
Możemy to zaobserwować, wyświetlając te trzy wartości z wysoką precyzją:

````python
print("0.1 = {0:.17f}".format(0.1))
print("0.2 = {0:.17f}".format(0.2))
print("0.3 = {0:.17f}".format(0.3))
````

Jesteśmy przyzwyczajeni do postrzegania liczb w systemie dziesiętnym (o podstawie 10), więc każdy ułamek musi być wyrażony jako suma potęg dziesięciu:
$$
1 /8 = 1\cdot 10^{-1} + 2\cdot 10^{-2} + 5\cdot 10^{-3}
$$
W znanym nam systemie dziesiętnym przedstawiamy to w znanym wyrażeniu dziesiętnym: $0,125$.

Komputery zazwyczaj przechowują wartości w notacji binarnej, tak że każda liczba jest wyrażona jako suma potęg 2:
$$
1/8 = 0\cdot 2^{-1} + 0\cdot 2^{-2} + 1\cdot 2^{-3}
$$
W notacji o podstawie 2 możemy zapisać to jako $0,001_2$, gdzie indeks dolny 2 oznacza notację binarną.
Wartość $0,125 = 0,001_2$ jest liczbą, którą zarówno zapis binarny, jak i dziesiętny mogą przedstawić przy użyciu skończonej liczby cyfr.

W znanym nam zapisie liczb w systemie dziesiętnym prawdopodobnie znasz liczby, których nie da się wyrazić przy użyciu skończonej liczby cyfr.
Na przykład, dzielenie $1$ przez $3$ daje, w standardowym zapisie dziesiętnym:
$$
1 / 3 = 0,333333333\cdots
$$
Trójki ciągną się w nieskończoność: to znaczy, aby naprawdę przedstawić ten iloraz, liczba potrzebnych cyfr jest nieskończona!

Podobnie istnieją liczby, dla których zapisy binarne wymagają nieskończonej liczby cyfr.
Na przykład:
$$
1 / 10 = 0,00011001100110011\cdots_2
$$
Tak jak notacja dziesiętna wymaga nieskończonej liczby cyfr, aby idealnie przedstawić $1/3$, tak notacja binarna wymaga nieskończonej liczby cyfr, aby przedstawić $1/10$.
W większości systemów Python wewnętrznie obcina te reprezentacje do 52 bitów poza pierwszym bitem niezerowym.
