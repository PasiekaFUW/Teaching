# Wbudowane struktury danych

Poznaliśmy już proste typy w Pythonie: ``int``, ``float``, ``complex``, ``bool``, ``str`` i tak dalej.
Python posiada również kilka wbudowanych typów złożonych, które działają jako kontenery dla innych typów.
Te typy złożone to:

| Nazwa typu | Przykład                  | Opis                                          |
|-----------|---------------------------|-----------------------------------------------|
| ``list``  | ``[1, 2, 3]``             | Uporządkowana kolekcja                        |
| ``tuple`` | ``(1, 2, 3)``             | Niemutowalna uporządkowana kolekcja           |
| ``dict``  | ``{'a':1, 'b':2, 'c':3}`` | Nieuporządkowane mapowanie (klucz, wartość)   |
| ``set``   | ``{1, 2, 3}``             | Nieuporządkowana kolekcja unikalnych wartości |

Jak widać, okrągłe, kwadratowe i klamrowe nawiasy mają różne znaczenia w zależności od rodzaju tworzonej kolekcji.

## Listy
Listy to podstawowy typ *uporządkowanej* i *mutowalnej* kolekcji danych w Pythonie.
Można je zdefiniować za pomocą wartości oddzielonych przecinkami wewnątrz nawiasów kwadratowych; oto na przykład lista kilku pierwszych liczb pierwszych:

```python
L = [2, 3, 5, 7]
```

Listy mają do dyspozycji wiele przydatnych właściwości i metod.
Przyjrzymy się tutaj szybko kilku najczęstszym i najbardziej przydatnym:

```python
# Długość listy
len(L)
```

```python
# Dołączenie wartości na końcu
L.append(11)
L
```

```python
#Dodawanie łączy listy (konkatenacja)
L + [13, 17, 19]
```

```python
# sort() sortuje elementy listy
L = [2, 5, 1, 6, 3, 4]
L.sort()
L
```

Ponadto istnieje znacznie więcej wbudowanych metod dla list; są one dobrze opisane w [dokumentacji online Pythona](https://docs.python.org/3/tutorial/datastructures.html).

Chociaż do tej pory pokazywaliśmy listy zawierające wartości jednego typu, jedną z potężnych cech złożonych obiektów w Pythonie jest to, że mogą one zawierać obiekty *dowolnego* typu, a nawet mieszankę typów. Na przykład:

```python
L = [1, 'two', 3.14, [0, 3, 5]]
```

Ta elastyczność jest konsekwencją dynamicznego systemu typów Pythona.
Tworzenie takiej mieszanej sekwencji w statycznie typowanym języku, takim jak C, może przyprawić o zawrót głowy!
Widzimy, że listy mogą nawet zawierać inne listy jako elementy.
Taka elastyczność typów jest kluczowym elementem tego, co sprawia, że kod w Pythonie pisze się stosunkowo szybko i łatwo.

Do tej pory rozważaliśmy manipulacje listami jako całością; kolejnym niezbędnym elementem jest dostęp do poszczególnych elementów.
W Pythonie odbywa się to za pomocą *indeksowania* (indexing) i *wycinania* (slicing), którym przyjrzymy się w następnej kolejności.

### Indeksowanie i wycinanie list
Python zapewnia dostęp do elementów w typach złożonych poprzez *indeksowanie* dla pojedynczych elementów i *wycinanie* dla wielu elementów.
Jak zobaczymy, oba te działania są oznaczane składnią nawiasów kwadratowych.
Załóżmy, że wracamy do naszej listy kilku pierwszych liczb pierwszych:

```python
L = [2, 3, 5, 7, 11]
```

Python używa indeksowania *od zera*, więc możemy uzyskać dostęp do pierwszego i drugiego elementu, używając następującej składni:

```python
L[0]
```

```python
L[1]
```

Dostęp do elementów na końcu listy można uzyskać za pomocą liczb ujemnych, zaczynając od -1:

```python
L[-1]
```

```python
L[-2]
```

Ten schemat indeksowania można zwizualizować za pomocą poniższej tabeli:

| Indeks dodatni | 0 | 1 | 2 | 3 | 4 |
| :--- | :---: | :---: | :---: | :---: | :---: |
| **Wartość na liście** | **2** | **3** | **5** | **7** | **11** |
| **Indeks ujemny** | -5 | -4 | -3 | -2 | -1 |

Tutaj wartości na liście są reprezentowane przez duże liczby w kwadratach; indeksy listy są reprezentowane przez małe liczby powyżej i poniżej.
W tym przypadku, ``L[2]`` zwraca ``5``, ponieważ jest to kolejna wartość pod indeksem ``2``.

O ile *indeksowanie* jest sposobem na pobranie pojedynczej wartości z listy, *wycinanie* to sposób na dostęp do wielu wartości w podlistach.
Używa się tu dwukropka, aby wskazać punkt początkowy (włącznie) i punkt końcowy (wyłącznie) wycinanej tablicy.
Na przykład, aby uzyskać pierwsze trzy elementy listy, możemy napisać:

```python
L[0:3]
```

Zwróć uwagę, gdzie na poprzednim diagramie znajdują się ``0`` i ``3``, i jak wycinek pobiera tylko wartości między tymi indeksami.
Jeśli pominiemy pierwszy indeks, przyjmowane jest domyślnie ``0``, więc możemy napisać równoważnie:

```python
L[:3]
```

Podobnie, jeśli pominiemy ostatni indeks, domyślnie przyjmuje on długość listy.
Tym samym dostęp do ostatnich trzech elementów można uzyskać w następujący sposób:

```python
L[-3:]
```

Na koniec, możliwe jest określenie trzeciej liczby całkowitej, która reprezentuje rozmiar kroku; na przykład, aby wybrać co drugi element listy, możemy napisać:

```python
L[::2]  # równoważne L[0:len(L):2]
```

Szczególnie przydatną wersją tego rozwiązania jest określenie ujemnego kroku, co odwróci tablicę:

```python
L[::-1]
```

Zarówno indeksowanie, jak i wycinanie mogą być używane do ustawiania elementów, a także do uzyskiwania do nich dostępu.
Składnia jest taka, jakiej można by się spodziewać:

```python
L[0] = 100
print(L)
```

```python
L[1:3] = [55, 56]
print(L)
```

Bardzo podobna składnia wycinania jest również używana w wielu pakietach zorientowanych na analizę danych, w tym w NumPy i Pandas (o których wspomniano we wstępie).

Teraz, gdy poznaliśmy listy w Pythonie i sposób dostępu do elementów w uporządkowanych typach złożonych, przyjrzyjmy się trzem pozostałym standardowym złożonym typom danych wymienionym wcześniej.

## Krotki (Tuples)
Krotki są pod wieloma względami podobne do list, ale definiuje się je za pomocą nawiasów okrągłych zamiast kwadratowych:

```python
t = (1, 2, 3)
```

Można je również zdefiniować w ogóle bez nawiasów:

```python
t = 1, 2, 3
print(t)
```

Podobnie jak omówione wcześniej listy, krotki mają swoją długość, a poszczególne elementy można wyodrębnić za pomocą indeksowania w nawiasach kwadratowych:

```python
len(t)
```

```python
t[0]
```

Główną cechą odróżniającą krotki jest to, że są *niemutowalne* (niezmienne): oznacza to, że po ich utworzeniu nie można zmienić ich rozmiaru ani zawartości:

```python
t[1] = 4
```

```python
t.append(4)
```

Krotki są bardzo często używane w programach w Pythonie; szczególnie powszechnym przypadkiem są funkcje, które zwracają wiele wartości.
Na przykład, metoda ``as_integer_ratio()`` dla obiektów zmiennoprzecinkowych zwraca licznik i mianownik; ta podwójna wartość jest zwracana właśnie w postaci krotki:

```python
x = 0.125
x.as_integer_ratio()
```

Te wielokrotne wartości mogą być przypisywane do zmiennych indywidualnie w następujący sposób:

```python
numerator, denominator = x.as_integer_ratio()
print(numerator / denominator)
```

Logika indeksowania i wycinania omówiona wcześniej dla list działa również w przypadku krotek, podobnie jak mnóstwo innych metod.
Odwołaj się do [dokumentacji Pythona online](https://docs.python.org/3/tutorial/datastructures.html), aby zapoznać się z ich pełniejszą listą.

## Słowniki (Dictionaries)
Słowniki są niezwykle elastycznymi mapowaniami kluczy do wartości i stanowią podstawę dużej części wewnętrznej implementacji Pythona.
Można je utworzyć za pomocą oddzielonej przecinkami listy par ``klucz:wartość`` wewnątrz nawiasów klamrowych:

```python
numbers = {'one':1, 'two':2, 'three':3}
```

Dostęp do elementów i ich ustawianie odbywa się za pomocą składni indeksowania używanej dla list i krotek, z tą różnicą, że tutaj indeksem nie jest liczba liczona od zera, ale prawidłowy klucz w słowniku:

```python
# Dostęp do wartości poprzez klucz
numbers['two']
```

Nowe elementy można również dodawać do słownika za pomocą indeksowania:

```python
# Ustawienie nowej pary klucz:wartość
numbers['ninety'] = 90
print(numbers)
```

Pamiętaj, że słowniki nie zachowują żadnego porządku (kolejności) dla wprowadzanych parametrów; jest to celowe działanie.
Brak uporządkowania pozwala na bardzo wydajną implementację słowników, dzięki czemu losowy dostęp do elementów jest błyskawiczny, niezależnie od rozmiaru słownika (jeśli jesteś ciekawy, jak to działa, poczytaj o koncepcji *tablicy mieszającej* - hash table).
[Dokumentacja Pythona](https://docs.python.org/3/library/stdtypes.html) zawiera kompletną listę metod dostępnych dla słowników.

## Zbiory (Sets)

Czwartą podstawową kolekcją jest zbiór, który zawiera nieuporządkowane kolekcje unikalnych elementów.
Definiuje się je podobnie jak listy i krotki, z tą różnicą, że używają nawiasów klamrowych znanych ze słowników:

```python
primes = {2, 3, 5, 7}
odds = {1, 3, 5, 7, 9}
```

Jeśli znasz matematykę zbiorów, na pewno kojarzysz operacje takie jak suma, część wspólna (iloczyn), różnica, różnica symetryczna i inne.
Zbiory w Pythonie mają wbudowane wszystkie te operacje poprzez metody lub operatory.
Dla każdej z nich pokażemy dwie równoważne formy zapisu:

```python
# suma (union): elementy występujące w którymkolwiek ze zbiorów
primes | odds      # za pomocą operatora
primes.union(odds) # to samo za pomocą metody
```

```python
# część wspólna (intersection): elementy występujące w obu zbiorach
primes & odds             # za pomocą operatora
primes.intersection(odds) # to samo za pomocą metody
```

```python
# różnica (difference): elementy w primes, ale nie w odds
primes - odds           # za pomocą operatora
primes.difference(odds) # to samo za pomocą metody
```

```python
# różnica symetryczna (symmetric difference): elementy występujące tylko w jednym zbiorze
primes ^ odds                     # za pomocą operatora
primes.symmetric_difference(odds) # to samo za pomocą metody
```

Dostępnych jest znacznie więcej metod i operacji na zbiorach.
Prawdopodobnie już zgadłeś, co powiem dalej: odwołaj się do [dokumentacji Pythona online](https://docs.python.org/3/library/stdtypes.html), aby uzyskać kompletne źródło wiedzy.

# Przepływ sterowania

*Przepływ sterowania* (control flow) to miejsce, gdzie w programowaniu zaczyna się prawdziwa akcja.
Bez niego program jest po prostu listą instrukcji wykonywanych sekwencyjnie.
Dzięki przepływowi sterowania możesz wykonywać określone bloki kodu warunkowo i/lub wielokrotnie: te podstawowe elementy budulcowe można łączyć w zaskakująco skomplikowane programy!

Omówimy tutaj *instrukcje warunkowe* (w tym "``if``", "``elif``" i "``else``") oraz *pętle* (w tym "``for``" i "``while``" oraz towarzyszące im "``break``", "``continue``" i "``pass``").

## Instrukcje warunkowe: ``if``-``elif``-``else``:
Instrukcje warunkowe, często nazywane instrukcjami *if-then* (jeśli-to), pozwalają programiście na wykonanie określonych fragmentów kodu w zależności od pewnego warunku logicznego (prawda/fałsz).
Oto podstawowy przykład instrukcji warunkowej w Pythonie:

```python
x = -15

if x == 0:
    print(x, "jest zerem")
elif x > 0:
    print(x, "jest dodatnie")
elif x < 0:
    print(x, "jest ujemne")
else:
    print(x, "jest czymś, czego jeszcze nie widziałem...")
```

Zwróć szczególną uwagę na użycie dwukropków (``:``) i białych znaków (wcięć) do oznaczania oddzielnych bloków kodu.

Python wykorzystuje instrukcje ``if`` i ``else``, które są często używane w innych językach; jego bardziej unikalnym słowem kluczowym jest ``elif``, czyli skrót od "else if" (w przeciwnym razie jeśli).
W tych klauzulach warunkowych bloki ``elif`` i ``else`` są opcjonalne; dodatkowo możesz włączyć tyle instrukcji ``elif``, ile tylko chcesz.

## Pętle ``for``
Pętle w Pythonie to sposób na wielokrotne wykonanie jakiejś instrukcji.
Jeśli więc na przykład chcielibyśmy wypisać każdy element z listy, możemy użyć pętli ``for``:

```python
for N in [2, 3, 5, 7]:
    print(N, end=' ') # wypisz wszystko w tej samej linii
```

Zauważ prostotę pętli ``for``: określamy zmienną, której chcemy użyć, sekwencję, po której chcemy iterować (przejść), i używamy operatora "``in``" (w), aby połączyć je w intuicyjny i czytelny sposób.
Mówiąc dokładniej, obiektem po prawej stronie "``in``" może być dowolny *iterator* Pythona.
Iterator można traktować jako uogólnioną sekwencję, o czym porozmawiamy w rozdziale [Iteratory](10-Iterators.ipynb).

Na przykład jednym z najczęściej używanych iteratorów w Pythonie jest obiekt ``range``, który generuje sekwencję liczb:

```python
for i in range(10):
    print(i, end=' ')
```

Zauważ, że zakres domyślnie zaczyna się od zera i że zgodnie z konwencją górna granica zakresu nie jest uwzględniana w wynikach.
Obiekty typu range mogą również przyjmować bardziej skomplikowane wartości:

```python
# zakres od 5 do 10 (bez 10)
list(range(5, 10))
```

```python
# zakres od 0 do 10 (bez 10) z krokiem 2
list(range(0, 10, 2))
```

Możesz zauważyć, że znaczenie argumentów ``range`` jest bardzo podobne do składni wycinania (slicing), którą omówiliśmy w sekcji *Listy*.

Zwróć uwagę, że zachowanie ``range()`` to jedna z różnic między Pythonem 2 a Pythonem 3: w Pythonie 2 ``range()`` generuje listę, podczas gdy w Pythonie 3 ``range()`` generuje obiekt iterowalny.

## Pętle ``while``
Innym rodzajem pętli w Pythonie jest pętla ``while`` (dopóki), która iteruje, dopóki nie zostanie spełniony pewien warunek:

```python
i = 0
while i < 10:
    print(i, end=' ')
    i += 1
```

Argument pętli ``while`` jest oceniany jako wyrażenie logiczne (Boolean), a pętla jest wykonywana, dopóki wyrażenie to nie zwróci wartości False (Fałsz).

## ``break`` i ``continue``: Dostrajanie Twoich pętli
Istnieją dwie przydatne instrukcje, których można użyć wewnątrz pętli, aby dostosować sposób ich wykonywania:

- Instrukcja ``break`` (przerwij) całkowicie przerywa działanie pętli.
- Instrukcja ``continue`` (kontynuuj) pomija resztę bieżącego przebiegu pętli i przechodzi do następnej iteracji.

Można ich używać zarówno w pętlach ``for``, jak i ``while``.

Oto przykład użycia ``continue`` do wypisania ciągu liczb nieparzystych.
W tym przypadku wynik można by równie dobrze osiągnąć za pomocą instrukcji ``if-else``, ale czasami instrukcja ``continue`` może być wygodniejszym sposobem na wyrażenie Twojego pomysłu:

```python
for n in range(20):
    # jeśli reszta z dzielenia n / 2 to 0, pomiń resztę pętli
    if n % 2 == 0:
        continue
    print(n, end=' ')
```

Oto przykład instrukcji ``break`` użytej do mniej banalnego zadania.
Ta pętla wypełni listę wszystkimi liczbami Fibonacciego aż do określonej wartości:

```python
a, b = 0, 1
amax = 1000
L = []

while True:
    (a, b) = (b, a + b)
    if a > amax:
        break
    L.append(a)

print(L)
```

Zauważ, że użyliśmy pętli ``while True`` (dopóki Prawda), która działałaby w nieskończoność, gdybyśmy nie zastosowali instrukcji ``break``!

## Pętle z blokiem ``else``
Jednym z rzadko używanych wzorców dostępnych w Pythonie jest instrukcja ``else`` jako część pętli ``for`` lub ``while``.
Omówiliśmy blok ``else`` wcześniej: wykonuje się on, jeśli wszystkie instrukcje ``if`` i ``elif`` zwracają ``False``.
Pętla-``else`` to prawdopodobnie jedna z najbardziej myląco nazwanych instrukcji w Pythonie; wolałbym myśleć o niej jako o instrukcji ``nobreak`` (brak przerwania): to znaczy blok ``else`` jest wykonywany tylko wtedy, gdy pętla kończy się naturalnie, bez napotkania instrukcji ``break``.

Jako przykład, gdzie może to być przydatne, rozważ poniższą (niezoptymalizowaną) implementację *Sita Eratostenesa*, znanego algorytmu znajdowania liczb pierwszych:

```python
L = []
nmax = 30

for n in range(2, nmax):
    for factor in L:
        if n % factor == 0:
            break
    else: # no break
        L.append(n)
print(L)
```

Instrukcja ``else`` wykonuje się tylko wtedy, gdy żaden z dzielników nie dzieli danej liczby.
Instrukcja ``else`` działa podobnie z pętlą ``while``.

# Definiowanie i używanie funkcji

Do tej pory nasze skrypty były prostymi blokami kodu jednorazowego użytku.
Jednym ze sposobów na uporządkowanie naszego kodu w Pythonie i uczynienie go bardziej czytelnym oraz przydatnym do ponownego użycia jest wyodrębnienie użytecznych elementów do postaci wielokrotnie używanych *funkcji*.
Omówimy tutaj dwa sposoby tworzenia funkcji: instrukcję ``def``, przydatną dla dowolnego typu funkcji, oraz instrukcję ``lambda``, przydatną do tworzenia krótkich funkcji anonimowych.

## Używanie funkcji

Funkcje to grupy kodu, które mają swoją nazwę i mogą być wywoływane za pomocą nawiasów okrągłych.
Widzieliśmy już funkcje wcześniej. Na przykład, ``print`` w Pythonie 3 jest funkcją:

```python
print('abc')
```

Tutaj ``print`` to nazwa funkcji, a ``'abc'`` to jej *argument*.

Oprócz argumentów zwykłych, istnieją *argumenty nazwane* (keyword arguments), które są określane przez nazwę.
Jednym z dostępnych argumentów nazwanych dla funkcji ``print()`` (w Pythonie 3) jest ``sep``, który informuje, jaki znak lub znaki powinny być użyte do oddzielenia wielu elementów:

```python
print(1, 2, 3)
```

```python
print(1, 2, 3, sep='--')
```

Gdy argumenty nienazwane są używane razem z argumentami nazwanymi, argumenty nazwane muszą znajdować się na końcu.

## Definiowanie funkcji
Funkcje stają się jeszcze bardziej użyteczne, gdy zaczynamy definiować własne, organizując funkcjonalności do wykorzystania w wielu miejscach.
W Pythonie funkcje definiuje się za pomocą instrukcji ``def``.
Na przykład, możemy ująć naszą wersję kodu ciągu Fibonacciego z poprzedniej sekcji w następujący sposób:

```python
def fibonacci(N):
    L = []
    a, b = 0, 1
    while len(L) < N:
        a, b = b, a + b
        L.append(a)
    return L
```

Teraz mamy funkcję o nazwie ``fibonacci``, która przyjmuje pojedynczy argument ``N``, wykonuje na nim pewne operacje i zwraca (``return``) wartość; w tym przypadku jest to lista pierwszych ``N`` liczb Fibonacciego:

```python
fibonacci(10)
```

Jeśli znasz języki o silnym typowaniu, takie jak ``C``, od razu zauważysz, że nie ma tu żadnych informacji o typie powiązanych z wejściami (argumentami) ani wyjściami funkcji.
Funkcje w Pythonie mogą zwracać dowolny obiekt Pythona, prosty lub złożony, co oznacza, że konstrukcje, które mogą być trudne w innych językach, w Pythonie są proste.

Na przykład, wiele zwracanych wartości jest po prostu umieszczanych w krotce (tuple), co oznaczają przecinki:

```python
def real_imag_conj(val):
    return val.real, val.imag, val.conjugate()

r, i, c = real_imag_conj(3 + 4j)
print(r, i, c)
```

## Wartości domyślne argumentów

Często definiując funkcję, mamy pewne wartości, których chcielibyśmy, aby funkcja używała przez *większość* czasu, ale chcielibyśmy też dać użytkownikowi pewną elastyczność.
W takim przypadku możemy użyć *wartości domyślnych* dla argumentów.
Rozważ funkcję ``fibonacci`` z wcześniejszej części.
Co jeśli chcielibyśmy, aby użytkownik mógł pobawić się wartościami początkowymi?
Możemy to zrobić w następujący sposób:

```python
def fibonacci(N, a=0, b=1):
    L = []
    while len(L) < N:
        a, b = b, a + b
        L.append(a)
    return L
```

W przypadku wywołania z jednym argumentem, wynik działania funkcji jest identyczny jak poprzednio:

```python
fibonacci(10)
```

Ale teraz możemy użyć funkcji do odkrywania nowych rzeczy, takich jak efekt nowych wartości początkowych:

```python
fibonacci(10, 0, 2)
```

Wartości można również określić po nazwie, jeśli jest to pożądane, w którym to przypadku kolejność podanych wartości nazwanych nie ma znaczenia:

```python
fibonacci(10, b=3, a=1)
```

## ``*args`` i ``**kwargs``: Elastyczne argumenty
Czasami możesz chcieć napisać funkcję, w której od początku nie wiesz, ile argumentów przekaże użytkownik.
W takim przypadku możesz użyć specjalnej formy ``*args`` i ``**kwargs``, aby przechwycić wszystkie przekazane argumenty.
Oto przykład:

```python
def catch_all(*args, **kwargs):
    print("args =", args)
    print("kwargs = ", kwargs)
```

```python
catch_all(1, 2, 3, a=4, b=5)
```

```python
catch_all('a', keyword=2)
```

Ważne są tutaj nie tyle same nazwy ``args`` i ``kwargs``, co poprzedzające je znaki ``*``.
``args`` i ``kwargs`` to po prostu nazwy zmiennych używane często z racji konwencji, jako skrót od "arguments" (argumenty) i "keyword arguments" (argumenty nazwane).
Kluczowa różnica polega na znakach gwiazdki: pojedyncza ``*`` przed zmienną oznacza "rozwiń to jako sekwencję", podczas gdy podwójna ``**`` przed zmienną oznacza "rozwiń to jako słownik".
W rzeczywistości tej składni można używać nie tylko przy definicji funkcji, ale także przy jej wywoływaniu!

```python
inputs = (1, 2, 3)
keywords = {'pi': 3.14}

catch_all(*inputs, **keywords)
```

## Funkcje anonimowe (``lambda``)
Wcześniej omówiliśmy pokrótce najczęstszy sposób definiowania funkcji, instrukcję ``def``.
Prawdopodobnie natkniesz się też na inny sposób definiowania krótkich, jednorazowych funkcji za pomocą instrukcji ``lambda``.
Wygląda to mniej więcej tak:

```python
add = lambda x, y: x + y
add(1, 2)
```

Ta funkcja lambda jest z grubsza odpowiednikiem:

```python
def add(x, y):
    return x + y
```

Po co więc w ogóle chcieć używać czegoś takiego?
Głównie sprowadza się to do faktu, że *wszystko jest obiektem* w Pythonie, nawet same funkcje!
Oznacza to, że funkcje mogą być przekazywane jako argumenty do innych funkcji.

Jako przykład, przypuśćmy, że mamy pewne dane zapisane w liście słowników:

```python
data = [{'first':'Guido', 'last':'Van Rossum', 'YOB':1956},
        {'first':'Grace', 'last':'Hopper',     'YOB':1906},
        {'first':'Alan',  'last':'Turing',     'YOB':1912}]
```

Teraz przypuśćmy, że chcemy posortować te dane.
Python posiada funkcję ``sorted``, która to robi:

```python
sorted([2,4,3,5,1,6])
```

Jednakże słowniki nie podlegają samodzielnie porządkowaniu: musimy znaleźć sposób, aby powiedzieć funkcji, *jak* ma posortować nasze dane.
Możemy to zrobić, określając funkcję ``key`` (klucz), czyli funkcję, która dla danego elementu zwraca klucz sortowania dla tego elementu:

```python
# sortuj alfabetycznie po imieniu (first name)
sorted(data, key=lambda item: item['first'])
```

```python
# sortuj po roku urodzenia (YOB)
sorted(data, key=lambda item: item['YOB'])
```

Chociaż te funkcje służące za klucz z pewnością mogłyby zostać utworzone przy użyciu standardowej składni ``def``, składnia ``lambda`` jest bardzo wygodna w przypadku takich krótkich, jednorazowych funkcji.

# Błędy i wyjątki

Niezależnie od Twoich umiejętności jako programista, prędzej czy później popełnisz jakiś błąd w kodzie.
Takie błędy występują w trzech podstawowych wariantach:

- *Błędy składniowe (Syntax errors):* Błędy, w których kod nie jest poprawnym kodem Pythona (zazwyczaj łatwe do naprawienia).
- *Błędy czasu wykonania (Runtime errors):* Błędy, w których poprawny składniowo kod nie może zostać wykonany, być może z powodu nieprawidłowych danych wprowadzonych przez użytkownika (czasami łatwe do naprawienia).
- *Błędy semantyczne (Semantic errors):* Błędy w logice: kod wykonuje się bez problemu, ale wynik nie jest taki, jakiego oczekujesz (często bardzo trudne do wyśledzenia i naprawienia).

Tutaj skupimy się na tym, jak zgrabnie radzić sobie z *błędami czasu wykonania*.
Jak zobaczymy, Python radzi sobie z nimi za pomocą platformy *obsługi wyjątków* (exception handling).

## Błędy czasu wykonania

Jeśli zaprogramowałeś już cokolwiek w Pythonie, prawdopodobnie natknąłeś się na błędy czasu wykonania.
Mogą one wystąpić na wiele sposobów.

Na przykład, jeśli spróbujesz odwołać się do niezdefiniowanej zmiennej:

```python
print(Q)
```

Albo jeśli spróbujesz wykonać operację, która nie jest zdefiniowana:

```python
1 + 'abc'
```

Albo być może próbujesz obliczyć matematycznie błędny wynik:

```python
2 / 0
```

Albo próbujesz uzyskać dostęp do elementu sekwencji, który nie istnieje:

```python
L = [1, 2, 3]
L[1000]
```

Zauważ, że w każdym z tych przypadków Python jest na tyle uprzejmy, że nie tylko informuje, iż wystąpił błąd, ale wyrzuca *sensowny* wyjątek, który zawiera informacje o tym, co dokładnie poszło nie tak, wraz z dokładną linią kodu, w której nastąpiła usterka.
Dostęp do takich precyzyjnych komunikatów jest niezwykle przydatny podczas próby wyśledzenia źródła problemów w kodzie.

## Przechwytywanie wyjątków: ``try`` i ``except``
Głównym narzędziem, które Python daje Ci do obsługi wyjątków w czasie wykonywania, jest klauzula ``try``...``except``.
Jej podstawowa struktura wygląda tak:

```python
try:
    print("to wykona się jako pierwsze")
except:
    print("to wykona się tylko w przypadku błędu")
```

Zauważ, że drugi blok w tym przypadku nie został wykonany: dzieje się tak dlatego, że pierwszy blok nie zwrócił błędu.
Wstawmy problematyczną instrukcję w bloku ``try`` i zobaczmy, co się stanie:

```python
try:
    print("spróbujmy czegoś:")
    x = 1 / 0 # ZeroDivisionError (błąd dzielenia przez zero)
except:
    print("stało się coś złego!")
```

Widzimy tutaj, że kiedy w instrukcji ``try`` został zgłoszony błąd (w tym przypadku ``ZeroDivisionError``), został on natychmiast przechwycony, a następnie wykonana została instrukcja ``except``.

Jednym ze sposobów, w jaki jest to często wykorzystywane, jest sprawdzanie danych wprowadzanych przez użytkownika wewnątrz funkcji lub innego fragmentu kodu.
Na przykład, możemy chcieć mieć funkcję, która wychwytuje dzielenie przez zero i zwraca bezpieczną wartość, być może odpowiednio dużą liczbę, taką jak $10^{100}$:

```python
def safe_divide(a, b):
    try:
        return a / b
    except:
        return 1E100
```

```python
safe_divide(1, 2)
```

```python
safe_divide(2, 0)
```

Z tym kodem wiąże się jednak pewien subtelny problem: co się stanie, gdy pojawi się inny rodzaj wyjątku? Na przykład, to prawdopodobnie nie jest efekt, o który nam chodziło:

```python
safe_divide (1, '2')
```

Dzielenie liczby całkowitej (int) i ciągu znaków (str) wywołuje błąd ``TypeError``, który nasz nadgorliwy kod przechwycił i błędnie uznał za ``ZeroDivisionError``!
Z tego powodu prawie zawsze lepszym pomysłem jest przechwytywanie wyjątków w sposób *jawny* (explicit):

```python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return 1E100
```

```python
safe_divide(1, 0)
```

```python
safe_divide(1, '2')
```

Teraz wychwytujemy tylko błędy dzielenia przez zero i przepuszczamy wszystkie inne błędy bez modyfikacji.

## Zgłaszanie wyjątków: ``raise``
Przekonaliśmy się, jak cenne jest posiadanie klarownych wyjątków podczas korzystania z elementów języka Python.
Równie cenne jest korzystanie z takich jasnych wyjątków we własnym kodzie, aby użytkownicy Twoich funkcji (a przede wszystkim Ty sam!) mogli łatwo dowiedzieć się, co spowodowało błąd.

Sposobem na wywołanie własnych wyjątków jest instrukcja ``raise`` (zgłoś). Na przykład:

```python
raise RuntimeError("mój własny komunikat o błędzie")
```

Jako przykład, gdzie może to być przydatne, wróćmy do naszej funkcji ``fibonacci``, którą zdefiniowaliśmy wcześniej:

```python
def fibonacci(N):
    L = []
    a, b = 0, 1
    while len(L) < N:
        a, b = b, a + b
        L.append(a)
    return L
```

Jednym z potencjalnych problemów w tym przypadku jest to, że wartość wejściowa mogłaby być ujemna.
Obecnie nie spowoduje to żadnego błędu w naszej funkcji, ale moglibyśmy chcieć poinformować użytkownika, że ujemne ``N`` po prostu nie jest obsługiwane.
Błędy wynikające z nieprawidłowych wartości parametrów zgodnie z konwencją prowadzą do zgłoszenia wyjątku ``ValueError``:

```python
def fibonacci(N):
    if N < 0:
        raise ValueError("N musi być nieujemne")
    L = []
    a, b = 0, 1
    while len(L) < N:
        a, b = b, a + b
        L.append(a)
    return L
```

```python
fibonacci(10)
```

```python
fibonacci(-10)
```

Teraz użytkownik wie dokładnie, dlaczego wprowadzone dane są nieprawidłowe, a nawet mógłby użyć bloku ``try``...``except``, aby sprawnie to obsłużyć!

```python
N = -10
try:
    print("próbujemy tego...")
    print(fibonacci(N))
except ValueError:
    print("Zła wartość: musisz zrobić coś innego")
```

## Głębsze spojrzenie na wyjątki

Chciałbym jeszcze krótko wspomnieć o kilku innych koncepcjach, na które możesz się natknąć.
Nie będę zagłębiał się w szczegóły tych pojęć ani w to, jak i dlaczego ich używać, lecz po prostu pokażę Ci składnię, abyś w razie potrzeby mógł zgłębić temat samodzielnie.

### Dostęp do komunikatu o błędzie

Czasami w instrukcji ``try``...``except`` chciałbyś móc pracować bezpośrednio z samym komunikatem błędu.
Można to zrobić za pomocą słowa kluczowego ``as``:

```python
try:
    x = 1 / 0
except ZeroDivisionError as err:
    print("Klasa błędu to:  ", type(err))
    print("Komunikat to:", err)
```

Dzięki temu wzorcowi możesz w większym stopniu dostosować i kontrolować obsługę wyjątków w swojej funkcji.

### Definiowanie własnych wyjątków
Oprócz wyjątków wbudowanych możliwe jest definiowanie własnych wyjątków niestandardowych poprzez *dziedziczenie klas* (class inheritance).
Na przykład, jeśli chcesz mieć specjalny rodzaj ``ValueError``, możesz zrobić to tak:

```python
class MySpecialError(ValueError):
    pass

raise MySpecialError("oto komunikat")
```

Umożliwiłoby to użycie bloku ``try``...``except``, który wychwytuje wyłącznie ten jeden, konkretny typ błędu:

```python
try:
    print("do something")
    raise MySpecialError("[informative error message here]")
except MySpecialError:
    print("do something else")
```

Z pewnością uznasz to za przydatne w miarę tworzenia bardziej zaawansowanego i zindywidualizowanego kodu.

## ``try``...``except``...``else``...``finally``
Oprócz ``try`` i ``except`` możesz użyć słów kluczowych ``else`` i ``finally``, aby jeszcze dokładniej dostroić sposób w jaki twój kod radzi sobie z awariami.
Podstawowa struktura wygląda tak:

```python
try:
    print("spróbuj czegoś tutaj")
except:
    print("to dzieje się tylko, gdy się nie uda (wystąpi błąd)")
else:
    print("to dzieje się tylko, gdy się uda (brak błędów)")
finally:
    print("to dzieje się zawsze, niezależnie od wszystkiego")
```

Przydatność klauzuli ``else`` jest tutaj jasna, ale jaki jest sens ``finally``?
Cóż, klauzula ``finally`` wykonuje się *naprawdę zawsze*: zazwyczaj widuję ją używaną do przeprowadzenia jakiegoś "sprzątania" (np. zamykania plików czy połączeń z bazą danych) po zakończeniu operacji, niezależnie od tego czy operacja ta zakończyła się sukcesem czy awarią.
